## Identify Top `true_wayfinder` Categories

### Subtask:
Get the list of the top 20 most frequent `true_wayfinder` categories from the DataFrame.


**Reasoning**:
The previous code block already identified and displayed the top 20 most frequent `true_wayfinder` categories. Now, I will store these categories in a list for future use, as specified in the subtask instructions.



```python
top_20_true_wayfinder_categories = df['true_wayfinder'].value_counts().head(20).index.tolist()
print("Top 20 true_wayfinder categories stored in a list:")
print(top_20_true_wayfinder_categories)
```

## Extract and Process Subjects per Category

### Subtask:
For each of the identified top `true_wayfinder` categories, filter the DataFrame, extract the `SUBJECT` column, split the subject strings (as they might contain multiple subjects separated by ';'), and count the occurrences of each individual subject. We will focus on the top N subjects for each `true_wayfinder` category.


**Reasoning**:
I will initialize a dictionary to store the top subjects for each category, then iterate through the top 20 `true_wayfinder` categories, filtering the DataFrame for each, processing the `SUBJECT` column by splitting and counting individual subjects, and finally storing the top 10 subjects for each category in the dictionary.



```python
top_subjects_per_category = {}

for category in top_20_true_wayfinder_categories:
    # Filter the DataFrame for the current true_wayfinder category
    category_df = df[df['true_wayfinder'] == category]

    # Select the SUBJECT column and drop NaN values
    subjects = category_df['SUBJECT'].dropna()

    if not subjects.empty:
        # Split subject strings by ';' and expand into individual subjects
        # First, convert to string to handle potential non-string types, then split
        split_subjects = subjects.astype(str).str.split(';').explode()

        # Clean up whitespace from each subject
        cleaned_subjects = split_subjects.str.strip()

        # Count the occurrences of each unique subject
        subject_counts = cleaned_subjects.value_counts()

        # Store the top 10 most frequent subjects for the current category
        top_subjects_per_category[category] = subject_counts.head(10)

print("Top subjects per category processed.")
# You can print the dictionary to inspect the results if needed:
# for category, subjects in top_subjects_per_category.items():
#     print(f"\nCategory: {category}")
#     print(subjects)
```

**Reasoning**:
Now that the top subjects for each `true_wayfinder` category have been extracted and processed, I will visualize these relationships using bar charts to make the findings easily interpretable.



```python
# @title
import matplotlib.pyplot as plt
import seaborn as sns

# Determine the number of categories to plot
num_categories = len(top_subjects_per_category)

# Calculate the number of rows and columns for subplots
# Aim for a layout that is easy to view, e.g., 4 columns
n_cols = 4
n_rows = (num_categories + n_cols - 1) // n_cols

plt.figure(figsize=(n_cols * 5, n_rows * 4))
plt.suptitle('Top 10 Subjects per true_wayfinder Category', y=1.02, fontsize=16)

for i, (category, subjects) in enumerate(top_subjects_per_category.items()):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.barplot(x=subjects.values, y=subjects.index, palette='viridis')
    plt.title(f'Category: {category}', fontsize=10)
    plt.xlabel('Count', fontsize=8)
    plt.ylabel('Subject', fontsize=8)
    plt.xticks(fontsize=7)
    plt.yticks(fontsize=7)
    plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap

plt.show()
```

**Reasoning**:

## Calculate Normalized Subject Frequencies

### Subtask:
For each `true_wayfinder` category, calculate the normalized frequency of its top subjects to understand their proportional prominence within each category.


**Reasoning**:
I will calculate the normalized frequency of top subjects for each `true_wayfinder` category by iterating through the previously generated `top_subjects_per_category` dictionary, summing the subject counts for each category, and then dividing each subject's count by that sum.



```python
normalized_top_subjects_per_category = {}

for category, subjects in top_subjects_per_category.items():
    total_subjects_in_category = subjects.sum()
    if total_subjects_in_category > 0:
        normalized_frequencies = subjects / total_subjects_in_category
        normalized_top_subjects_per_category[category] = normalized_frequencies
    else:
        normalized_top_subjects_per_category[category] = pd.Series(dtype='float64') # Handle empty series case

print("Normalized subject frequencies per category calculated.")
# You can print the dictionary to inspect the results if needed:
# for category, norm_subjects in normalized_top_subjects_per_category.items():
#     print(f"\nCategory: {category}")
#     print(norm_subjects)
```

## Visualize Normalized Subject Frequencies

### Subtask:
Create visualizations (e.g., bar charts) to display the normalized subject frequencies, making it easier to compare subject prominence across different `true_wayfinder` categories.


**Reasoning**:
Now that the normalized subject frequencies have been calculated for each `true_wayfinder` category, I will visualize these normalized frequencies using horizontal bar charts, similar to the previous visualization but with normalized values, to compare subject prominence more effectively across categories.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Determine the number of categories to plot
num_categories = len(normalized_top_subjects_per_category)

# Calculate the number of rows and columns for subplots, aiming for 4 columns
n_cols = 4
n_rows = (num_categories + n_cols - 1) // n_cols

plt.figure(figsize=(n_cols * 6, n_rows * 5)) # Adjusted figure size for better readability
plt.suptitle('Normalized Top 10 Subjects per true_wayfinder Category', y=1.02, fontsize=16)

for i, (category, norm_subjects) in enumerate(normalized_top_subjects_per_category.items()):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.barplot(x=norm_subjects.values, y=norm_subjects.index, hue=norm_subjects.index, palette='viridis', legend=False)
    plt.title(f'Category: {category}', fontsize=10)
    plt.xlabel('Normalized Frequency', fontsize=8) # Updated x-axis label
    plt.ylabel('Subject', fontsize=8)
    plt.xticks(fontsize=7)
    plt.yticks(fontsize=7)
    plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap

plt.show()
```

## Filter Data for Non-Fiction and Recent Additions

### Subtask:
Filter the dataset to include only non-fiction records and those cataloged after December 1, 2024, to prepare for the predictive modeling phase.


**Reasoning**:
I will convert the 'CATALOG DATE' column to datetime, filter for non-fiction records, then filter for dates after December 1, 2024, drop any invalid date entries, and finally display the head and info of the resulting DataFrame to verify the filtering.



```python
import pandas as pd

# 1. Convert 'CATALOG DATE' to datetime objects
df['CATALOG DATE'] = pd.to_datetime(df['CATALOG DATE'], errors='coerce')
print("CATALOG DATE column converted to datetime.")

# 2. Filter for non-fiction records
# Assuming 'True' in 'fiction' column denotes fiction, so we keep records where 'fiction' is not 'True'
non_fiction_df = df[df['fiction'] != 'True'].copy()
print(f"Filtered for non-fiction records: {len(non_fiction_df)} entries.")

# 3. Further filter for records with 'CATALOG DATE' after December 1, 2024
comparison_date = pd.to_datetime('2024-12-01')
recent_non_fiction_df = non_fiction_df[non_fiction_df['CATALOG DATE'] > comparison_date].copy()
print(f"Further filtered for dates after 2024-12-01: {len(recent_non_fiction_df)} entries.")

# 4. Drop any rows where 'CATALOG DATE' became NaT due to parsing errors
final_filtered_df = recent_non_fiction_df.dropna(subset=['CATALOG DATE']).copy()
print(f"Dropped NaT dates: {len(final_filtered_df)} entries remaining.")

# 5. Display the first few rows and the information of the final filtered DataFrame
print("\nFirst 5 rows of the final filtered DataFrame:")
print(final_filtered_df.head())

print("\nInfo of the final filtered DataFrame:")
final_filtered_df.info()
```
